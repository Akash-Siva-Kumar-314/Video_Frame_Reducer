```python
# Update these paths to point to your actual video location in Drive
input_video_path = '/content/drive/MyDrive/Video_Frame_Reducer/Cars_On_Road.mp4'
output_video_path = '/content/drive/MyDrive/Video_Frame_Reducer/your_reduced_frame_video.mp4'
```

```python
import cv2
```

```python
# Code for skipping 5 frames
cap = cv2.VideoCapture(input_video_path)

if not cap.isOpened():
    print("Error: Could not open video. Check the file path.")
else:
    # Get original video properties
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    original_fps = cap.get(cv2.CAP_PROP_FPS)

    # Define the codec and create a VideoWriter object
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')

    # Set the output FPS
    # Dividing the original FPS by 5 keeps the video duration the same.
    # (If you want a fast-forward/time-lapse effect instead, just use 'original_fps') -what!?
    output_fps = original_fps / 5

    out = cv2.VideoWriter(output_video_path, fourcc, output_fps, (width, height))
    #out = cv2.VideoWriter(output_video_path, fourcc, original_fps, (width, height))

    frame_count = 0
    saved_count = 0

    print("Processing video...")

    while True:
        ret, frame = cap.read()

        # Break the loop if we have reached the end of the video
        if not ret:
            break

        # Save every 5th frame (e.g., 0, 5, 10, 15...)
        if frame_count % 5 == 0:
            out.write(frame)
            saved_count += 1

        frame_count += 1

    # Release resources once finished
    cap.release()
    out.release()

    print(f"Done! Saved {saved_count} frames (from {frame_count} total frames).")
    print(f"Output saved to: {output_video_path}")
```

```python
'''
Note:
The Playback Speed: By default, this script divides the original frames per second (FPS) by 5.
This ensures the output video plays at normal speed but looks "choppy."
If you want the video to play 5 times faster (like a time-lapse), change output_fps = original_fps / 5 to just output_fps = original_fps.

or

out = cv2.VideoWriter(output_video_path, fourcc, output_fps, (width, height)) # normal speed but choppy
out = cv2.VideoWriter(output_video_path, fourcc, original_fps, (width, height)) # time lapse effect
'''
```

```python
# To print video information on frame, fps and duration
def print_video_stats(video_path, label):
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print(f"Error: Could not open {label} video at {video_path}")
        return

    # Get frame count and FPS
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)

    # Calculate duration (Frames / Frames Per Second = Seconds)
    duration_sec = frame_count / fps if fps > 0 else 0

    print(f"--- {label} Video ---")
    print(f"Total Frames: {frame_count}")
    print(f"FPS:          {fps:.2f}")
    print(f"Duration:     {duration_sec:.2f} seconds\n")

    cap.release()

# Print stats for both videos
print_video_stats(input_video_path, "Original")
print_video_stats(output_video_path, "Reduced")
```
